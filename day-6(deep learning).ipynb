```python
import tensorflow as tf
```

# **1. Creating tensors from constants**
*  you can create tensors using the if constant() function.

```python
# creaating a scalar(0D tensors)
scalar = tf.constant(42)
print(scalar)

```

```python
# Creating a vector (1D tensors)
vector = tf.constant([1,2,3,4])
print(vector)
```

```python
#creating a matrix (2D tensors)
matrix = tf.constant([[1,2],[3,4]])
print(matrix)
```

```python
# Creating a 3 tensor
tensor_3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(tensor_3d)
```

```python
variable_tensor = tf.Variable([[1.0, 2.0]])
```

```python
# Creating a tensor filled with zeros
zeros_tensor = tf.zeros((2, 3))

# Creating a tensor fille with ones
ones_tensor = tf.ones((3, 3))

# Creating a tensor filled with a specific value
filled_tensor = tf.fill((2, 2), 9)

print(zeros_tensor)
print(ones_tensor)
print(filled_tensor)
```

```python
#Creating a sequence from 0 to 9
range_tensor = tf.range(10)

#Creating a sequence of 5 values from 0 to 1
linspace_tensor = tf.linspace(0.0, 1.0, 5)

print(range_tensor)
print(linspace_tensor)
```

```python
# Creating a tensor with random values from a normal distribution
random_normal = tf.random.normal((2, 3), mean =0, stddev =1)

# Creating a tensor with random values from a uniform distribution
random_uniform = tf.random.uniform((2, 3), minval =0, maxval=10)

print(random_normal)
print(random_uniform)
```

```python
import numpy as np

#Creating a NumPy array
numpy_array = np.array([[1,2,3],[4,5,6]])

# Converting to a TensorFlow tensor
tensor_from_numpy = tf.convert_to_tensor(numpy_array)

print(tensor_from_numpy)
```

```python
tensor = tf.constant([[1, 2, 3], [4, 5, 6]])

print("Shape:", tensor.shape)
print("Data type:", tensor.dtype)
print("Rank (Dimensions):", tf.rank(tensor))
```
