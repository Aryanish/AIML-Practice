```python
import pandas as pd

def hex_to_rgb(hex_code):
    """Converts a hexadecimal color code to an RGB tuple.
    Handles empty strings and non-hexadecimal values by returning (0, 0, 0).
    """
    hex_code = hex_code.lstrip('#')
    if not hex_code:  # Check if the hex code is empty after removing '#'
        return (0, 0, 0)  # Return (0, 0, 0) for empty or invalid hex codes
    try:
        return tuple(int(hex_code[i:i+2], 16) for i in (0, 2, 4))
    except ValueError:
        return (0, 0, 0)  # Return (0, 0, 0) for invalid hex codes


# Apply the function to create separate RGB columns from Hex_Code
data[['Hex_Red', 'Hex_Green', 'Hex_Blue']] = data['Hex_Code'].apply(lambda x: pd.Series(hex_to_rgb(x)))

# Display the updated dataset
print(data.head())
```
