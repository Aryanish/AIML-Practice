```python
 import sqlite3
 # Connect to the database (or create it if it doesn't exist)
 conn = sqlite3.connect('store.db')
 cursor = conn.cursor()
 # Create the "products" table
 cursor.execute('''
    CREATE TABLE IF NOT EXISTS products (
        product_id INTEGER PRIMARY KEY,
        product_name TEXT,
        price REAL,
        quantity INTEGER
    )
 ''')
 # Commit the changes and close the connection
 conn.commit()
 conn.close()
import sqlite3
 conn = sqlite3.connect('store.db')
 cursor = conn.cursor()
 # Insert data into the "products" table
 cursor.execute('''
    INSERT INTO products (product_name, price, quantity)
    VALUES ('Laptop', 800.50, 10)
 ''')
 conn.commit()
 conn.close()
 #c) Adding a "description" column to the "products" table:
 import sqlite3
 conn = sqlite3.connect('store.db')
 cursor = conn.cursor()
 # Add the "description" column
 cursor.execute('''
    ALTER TABLE products
    ADD COLUMN description TEXT
 ''')
 conn.commit()
 conn.close()
 #d) Counting the number of products in the "products" table:
 import sqlite3
 conn = sqlite3.connect('store.db')
 cursor = conn.cursor()
 # Count the number of products
 cursor.execute("SELECT COUNT(*) FROM products")
 count = cursor.fetchone()[0]
 print(f"There are {count} products in the table.")
 conn.close()

 #e) Retrieving products with a price greater than 500:
 import sqlite3
 conn = sqlite3.connect('store.db')
 cursor =conn.cursor()
 # Retrieve products with price > 500
 cursor.execute(
 "SELECT * FROM products WHERE price > 500")
 products =cursor.fetchall()
 for product in products:
  print(product)
 conn.close()
```
