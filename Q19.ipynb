```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
# Define the documents
documents = [
    "The quick brown fox jumps over the lazy dog.",
    "The fox is quick and jumps high.",
    "A lazy dog lies under the tree."
]
```

```python

# Initialize the TfidfVectorizer
vectorizer = TfidfVectorizer()

# Calculate the TF-IDF matrix
tfidf_matrix = vectorizer.fit_transform(documents)

# Get feature names (words)
feature_names = vectorizer.get_feature_names_out()

# Convert the matrix to a DataFrame for readability
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)

# Print the TF-IDF values for each word in each document
print(tfidf_df)
```

```python
import tensorflow as tf
import numpy as np
```

```python
# Define the quadratic function: y = 2x^2 + 3x + 4
def quadratic_function(x):
    return 2 * x**2 + 3 * x + 4

# Generate synthetic data
x_data = np.linspace(-10, 10, 100)  # 100 points from -10 to 10
y_data = quadratic_function(x_data)
```

```python

# Define a simple quadratic regression model
class QuadraticModel(tf.keras.Model):
    def __init__(self):
        super(QuadraticModel, self).__init__()
        # Randomly initialized weights and bias
        self.w1 = tf.Variable(tf.random.normal([1]), name='w1')
        self.w2 = tf.Variable(tf.random.normal([1]), name='w2')
        self.b = tf.Variable(tf.random.normal([1]), name='b')

    def call(self, inputs):
        # Quadratic relationship: output = w1 * x^2 + w2 * x + b
        return self.w1 * inputs**2 + self.w2 * inputs + self.b
```

```python
# Instantiate the model
model = QuadraticModel()

# Define the loss function (Mean Squared Error) and optimizer
loss_fn = tf.keras.losses.MeanSquaredError()
optimizer = tf.keras.optimizers.Adam(learning_rate=0.01)
```

```python

# Training loop
for epoch in range(1000):  # Train for 1000 epochs
    with tf.GradientTape() as tape:
        predictions = model(x_data)
        loss = loss_fn(y_data, predictions)
    gradients = tape.gradient(loss, [model.w1, model.w2, model.b])
    optimizer.apply_gradients(zip(gradients, [model.w1, model.w2, model.b]))

    if (epoch + 1) % 100 == 0:  # Print loss every 100 epochs
        print(f"Epoch {epoch + 1}: Loss = {loss.numpy()}")
```

```python
# Print the learned weights and bias
print(f"Learned w1: {model.w1.numpy()[0]}")
print(f"Learned w2: {model.w2.numpy()[0]}")
print(f"Learned b: {model.b.numpy()[0]}")
```
