```python
#Wrtie a program to read a name from a user and convert all the alternate case of the name
```

```python
def alternate_case(name):
    result = ""
    for i, char in enumerate(name):
        if i % 2 == 0:
            result += char.lower()
        else:
            result += char.upper()
    return result

name = input("Enter your name: ")

converted_name = alternate_case(name)

print("Name with alternate case:", converted_name)

```

```python
def next_alphabet(string):
    result = ""
    for char in string:
        if char.isalpha():
            if char == 'z':
                result += 'a'
            elif char == 'Z':
                result += 'A'
            else:
                result += chr(ord(char) + 1)
        else:
            result += char
    return result

input_string = input("Enter a string: ")

output_string = next_alphabet(input_string)

print("String with next alphabets:", output_string)
```
