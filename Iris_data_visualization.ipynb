```python
# Begin by importing pandas and seaborn, a data analysis tookit and graphing library, respectively.

import pandas as pd

# To ignore warnings, use the following code to make the display more attractive.
# Import seaborn and matplotlib.
import warnings
warnings.filterwarnings("ignore")
import seaborn as sns
import matplotlib.pyplot as plt
sns.set(style="white", color_codes=True)

#To import the Iris dataset:
iris = pd.read_csv("Iris.csv") # the iris dataset is now a Pandas DataFrame

#To view Iris data below:
iris.head()
```

```python
# Samples from each species
iris["Species"].value_counts()
```

```python
# The pandas plot extenstion can be used to make a scatterplot
# Display your plot with plt.show

iris.plot(kind="scatter", x="SepalLengthCm", y="SepalWidthCm")
plt.show()
```

```python
#To change color and size, add the following:
iris.plot(kind="scatter", x="SepalLengthCm", y="SepalWidthCm",color="green",s=70 )
plt.show()
```

```python
# Use seaborn jointplot, to make bivariate scatterplots and univariate histograms in one graph
sns.jointplot(x="SepalLengthCm", y="SepalWidthCm", data=iris, size=5)
plt.show()
```

```python
# Modify the graph above by assigning each species an individual color.
sns.FacetGrid(iris, hue="Species", height=5) \
   .map(plt.scatter, "SepalLengthCm", "SepalWidthCm") \
   .add_legend()
plt.show()
```

```python
KS = {'color': ['blue', 'red', 'yellow']}
sns.FacetGrid(iris, hue_kws=KS, hue="Species", height=5) \
   .map(plt.scatter, "SepalLengthCm", "SepalWidthCm") \
   .add_legend()
plt.show()
```

```python
# To plot the species data using a box plot:

sns.boxplot(x="Species", y="PetalLengthCm", data=iris )
plt.show()
```

```python
ax= sns.boxplot(x="Species", y="PetalLengthCm", data=iris)
ax= sns.stripplot(x="Species", y="PetalLengthCm", data=iris, jitter=True, edgecolor="gray")
plt.show()
```

```python
ax= sns.boxplot(x="Species", y="PetalLengthCm", data=iris)
ax= sns.stripplot(x="Species", y="PetalLengthCm", data=iris, jitter=True, edgecolor="gray")

# Assuming the order of boxes is consistent with the order of species in the data:
# If your species order is different, you may need to adjust the indices accordingly.
try:
  # Attempt to access and modify the artists for each species
  boxtwo = ax.artists[2]  # Artist for the third species (index 2)
  boxtwo.set_facecolor('red')  # Set color to red
  boxtwo.set_edgecolor('black') # Set edge color to black

  boxthree = ax.artists[1]  # Artist for the second species (index 1)
  boxthree.set_facecolor('yellow')  # Set color to yellow
  boxthree.set_edgecolor('black') # Set edge color to black

except IndexError:
  print("Warning: Not enough boxes to color as expected.")

plt.show()
```

```python
#To change the fill color of the violin, choose desired colors and set equal to pallete

sns.violinplot(x="Species", y="PetalLengthCm",  palette={"blue","red","yellow"}, data=iris) # Remove size=6
plt.show()
```
