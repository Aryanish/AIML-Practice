```python
import pandas as pd
import seaborn as sns

# Load the tips dataset
tips = sns.load_dataset('tips')
```

```python
# 1. Average billing of the restaurant
average_billing = tips['total_bill'].mean()
print(f"Average billing: {average_billing}")
```

```python
# 2. Median price of total_bill
median_total_bill = tips['total_bill'].median()
print(f"Median total bill: {median_total_bill}")
```

```python
# 3. Most common tip size
most_common_tip = tips['tip'].mode()[0]
print(f"Most common tip size: {most_common_tip}")
```

```python
# 4. Variance of tips
variance_of_tips = tips['tip'].var()
print(f"Variance of tips: {variance_of_tips}")
```

```python
# 5. Standard deviation of total_bill
std_dev_total_bill = tips['total_bill'].std()
print(f"Standard deviation of total_bill: {std_dev_total_bill}")
```
