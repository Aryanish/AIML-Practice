# **SQLite**

SQLite is a lightweight, serverless, self-contained, zero-configuration, transactional SQL database engine.

It's the most widely deployed database engine in the world, embedded in:
1. Mobile apps (Android, iOS)
2. Web browsers (Chrome, Firefox, Safari)
3. Operating systems (Windows, macOS, Linux)
4. IoT devices

### Key characteristics:

1. Single ordinary disk file contains the entire database
2. No separate server process needed
3. Cross-platform (works on all major OS)
4. Public domain (completely free to use)

### Features & Advantages

1. Zero Configuration: No setup or administration needed
2. Serverless: No separate server process to install/configure
3. Single Database File: Entire database stored in one file
4. Compact: Library size <1MB>, minimal memory footprint
5. ACID Compliant Atomic, Consistent, Isolated, Durable transactions
6. Full-featured SQL: Supports most SQL92 features
7. Cross-platform: Works on all major operating systems
8. Public Domain: No licensing restrictions

### Installing SQLite Windows:
1. Download precompiled binaries from sqlite.org
2. Get the "sqlite-tools-win32-x86-*.zip" package
3. Extract to a folder (e.g., C:\sqlite)
4. Add to PATH or run from that directory

# **Connecting to SQLite Database using Python**

    import sqlite3

Create a database connection (if the file doesn't exist, it will be created)

    conn = sqlite3.connect("my_database.db")

Create a cursor object

    cursor = conn.cursor()

Print connection success message

    print("Connected to SQLite database successfully!")

Close the connection

    conn.close()

Connected to SQLite database successfully

#Creating a Table

    conn = sqlite3.connect("my_database.db")
    cursor = conn.cursor()

Create a table

    cursor.execute(""" CREATE TABLE IF NOT EXISTS users (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT NOT NULL,
        age INTEGER NOT NULL
    )
    """)

    conn.commit()
    conn.close()
    print("Table created successfully!")
Table created successfully!
