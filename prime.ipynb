```python
def is_prime(num):
  """
  Checks if a number is prime.

  Args:
    num: The number to check.

  Returns:
    True if the number is prime, False otherwise.
  """
  if num <= 1:
    return False
  for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
      return False
  return True

# Get input from the user
num = int(input("Enter a number: "))

# Check if the number is prime and print the result
if is_prime(num):
  print(f"{num} is a prime number.")
else:
  print(f"{num} is not a prime number.")
```

```python
import random
def fitness(route, distances):
  return sum (distance[route[i-1]][route[i]]for i in range(len(route)))
```

```python
def mutate(route):
  idx1, idx2 = random.sample(range(len(route)), 2)
  route[idx1], route[idx2] = route[idx2], route[idx1]
  return route
```

```python
distance = {
    'A':{'A':0, 'B':2, 'C':9, 'D':10, 'E': 4},
    'B':{'A':2, 'B':0, 'C':7, 'D':8, 'E': 1},
    'C':{'A':9, 'B':7, 'C':0, 'D':6, 'E': 5},
    'D':{'A':10, 'B':8, 'C':6, 'D':0, 'E': 3},
    'E':{'A':4, 'B':1, 'C':5, 'D':3, 'E': 0}
}
```

```python
initial_route = ['A', 'B', 'C', 'D', 'E']
best_route = initial_route
best_fitness = fitness(best_route, distance)

for _ in range(100):
  new_route = mutate(best_route[:])
  new_fitness = fitness(new_route, distance)

  if new_fitness < best_fitness:
    best_route = new_route
    best_fitness = new_fitness

print("Optimized Route:", best_route)
print("Distance:", best_fitness)
```
