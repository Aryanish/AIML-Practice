```python
nltk.download ('punkt_tab')
nltk.download ('stopwords')
```

```python
text = "Natural language processing is an amazinf field!"
```

```python
tokens =word_tokenize(text)
```

```python
stop_words = set(stopwords.words('english'))
firtered_tokens = [word for word in tokens if word.lower() not in stop_words]
```

```python
print("original Tokens:",tokens)
print("Filtered Tokens:",firtered_tokens)
```
