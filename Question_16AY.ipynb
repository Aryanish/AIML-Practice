```r
# Load necessary libraries
library(dplyr)
library(ggplot2)

# Example dataset
data <- data.frame(
  year = c(2025, 2025, 2025, 2024, 2025, 2024, 2025),
  month = c(1, 2, 3, 1, 4, 2, 5),
  sales = c(1000, 1500, 1200, 1300, 1800, 1100, 1600),
  category = c("Electronics", "Furniture", "Furniture", "Office Supplies", "Electronics", "Office Supplies", "Furniture")
)

# 1. Filter the data to only include the year 2025
data_2025 <- data %>% filter(year == 2025)

# 2. Group the data by category and calculate the average sales per category
avg_sales_by_category <- data_2025 %>%
  group_by(category) %>%
  summarize(avg_sales = mean(sales))

# 3. Plot a bar chart showing the average sales per category
ggplot(avg_sales_by_category, aes(x = category, y = avg_sales, fill = category)) +
  geom_bar(stat = "identity") +
  labs(title = "Average Sales per Category in 2025", x = "Category", y = "Average Sales") +
  theme_minimal()
```
