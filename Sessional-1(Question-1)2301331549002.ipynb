```python
import pandas as pd
import numpy as np

# data visualization
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.datasets import fetch_california_housing
```

```python
california_housing = fetch_california_housing()
df = pd.DataFrame(california_housing.data, columns=california_housing.feature_names)

# Add the target variable (median house value)
df['MedHouseVal'] = california_housing.target

# Total number of districts
total_districts = df.shape[0]
print(f"Total number of districts: {total_districts}")
```

```python
# Average median house value
average_med_house_value = df['MedHouseVal'].mean()
print(f"Average median house value across all districts: ${average_med_house_value:,.2f}")
```

```python
# District with highest median income
highest_med_inc_district = df.loc[df['MedInc'].idxmax()]
print("District with highest median income:")
print(highest_med_inc_district)
```

```python
# Filter districts where the median house value is above $300,000
high_value_districts = df[df['MedHouseVal'] > 300000]

# Average house age in those districts
average_house_age_high_value = high_value_districts['HouseAge'].mean()
print(f"Average house age in districts with median house value above $300,000: {average_house_age_high_value:.2f} years")
```

```python
import matplotlib.pyplot as plt

# Plot histogram of median house values
plt.figure(figsize=(10, 6))
plt.hist(df['MedHouseVal'], bins=50, color='skyblue', edgecolor='black')
plt.title('Distribution of Median House Values')
plt.xlabel('Median House Value ($)')
plt.ylabel('Number of Districts')
plt.grid(True)
plt.show()
```
