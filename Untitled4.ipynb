```python
import pybullet as p
import pybullet_data
import numpy as np
import cv2
from IPython.display import display, clear_output
```

```python
def display_image():
  """Capture and display the simulation image."""
  width, hight, rgb, depth,_=p.getCameraImage(
      width=640, height=480,
      viewMatrix = p.computerViewMatrixFromYawPitchRoll(
          cameraTargetPosition = [0, 0, 0.5],
          distance =2,
          yaw=90,
          pitch=-30,
          roll=0,
          upAxisIndex=2
      ),
      projectionMatrix=p.computeProjectionMatrixFOV(
          fov=60,
          aspect=640/480,
          nearVal=0.1,
          farVal=100
      )
  )
```
