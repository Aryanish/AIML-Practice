```python
import pandas as pd
```

```python
df = pd.read_csv('100 Sales Records.csv')
df.head()
```

```python
# Convert 'Order Date' and 'Ship Date' columns to datetime format
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python

# Calculate total revenue per region
total_revenue_per_region = df.groupby("Region")["Total Revenue"].sum()
```

```python

# Filter transactions where total profit is greater than 5000
high_profit_orders = df[df["Total Profit"] > 5000]
```

```python
# Display results
print("Total Revenue per Region:")
print(total_revenue_per_region)
print("\nHigh Profit Orders:")
print(high_profit_orders)
```
